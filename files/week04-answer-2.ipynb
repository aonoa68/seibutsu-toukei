```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```
# Equivalent to read.csv in R
url = 'https://raw.githubusercontent.com/aonoa68/Rdata/main/Animals.csv'
animals = pd.read_csv(url, index_col=0)  # row.names = 1 in R
print('Shape:', animals.shape)
```

```
animals.head()
```

```
# Equivalent to summary() in R
animals.describe(include='all')
```

```
animals
```

```
print('SD(body) =', animals['body'].std())
print('SD(brain) =', animals['brain'].std())
```

```
fig, ax = plt.subplots(figsize=(6,4))
for ord_ in animals['order'].unique():
    subset = animals[animals['order'] == ord_]
    ax.scatter(np.log(subset['body']), np.log(subset['brain']), label=ord_)
ax.set_title('Scatterplot of body and brain (log‑log scale)')
ax.set_xlabel('log(body kg)')
ax.set_ylabel('log(brain g)')
ax.legend(title='order', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
```

```
fig, ax = plt.subplots(figsize=(8,6))
ax.scatter(np.log(animals['body']), np.log(animals['brain']))
for idx, row in animals.iterrows():
    ax.text(np.log(row['body']), np.log(row['brain']), idx)
ax.set_title('Scatterplot with labels (log‑log scale)')
ax.set_xlabel('log(body kg)')
ax.set_ylabel('log(brain g)')
plt.tight_layout()
```
